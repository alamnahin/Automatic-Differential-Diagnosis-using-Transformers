```python
import pandas as pd
import os
import json
import ast
```

# converting release_conditions.json file to dataframe

```python
with open("release_conditions.json") as jf:
    data = json.load(jf)
```

```python
# keys present in the json data
keys = list(data.keys())
print('Dictionary Keys (all conditions): ',keys)
print()
print('Total Number of conditions: ',len(keys))
print()
print('Keys for a condition: ', data[keys[0]].keys())
```

```python
df = pd.DataFrame.from_dict(data, orient='index')
df = df.reset_index(drop=True)
```

```python
# save dataframe as csv
df.to_csv('release_conditions.csv', index=False)
```

## converting release_evidences.json to dataframe

```python
with open("release_evidences.json") as jf:
    data = json.load(jf)
```

```python
# keys present in the json data
keys = list(data.keys())
print('Dictionary Keys (all evidences): ',keys)
print()
print('Total Number of evidences: ',len(keys))
print()
print('Keys for an evidence: ', data[keys[0]].keys())
```

```python
df = pd.DataFrame.from_dict(data, orient='index')
df = df.reset_index(drop=True)
```

```python
# save dataframe as csv
df.to_csv('release_evidences.csv', index=False)
```

### separating unique values from the 'value meanings' column (used for multiple choice questions) from 'release_evidences'

```python
value_list = []
for value in df['value_meaning']:
    # check if the cell has values
    if value:
        value_list.append(value)
```

```python
col_names = ['index', 'fr', 'en']

values_df = pd.DataFrame(columns = col_names)

for value_dict in value_list:
    temp_df = pd.DataFrame.from_dict(value_dict, orient='index')
    temp_df = temp_df.reset_index()
    
    values_df = pd.concat([values_df, temp_df], ignore_index=True)
```

```python
len(values_df)
```

```python
# remove duplicate instances
unique_values_df = values_df.drop_duplicates(subset='index', keep='first')
len(unique_values_df)
```

```python
unique_values_df.reset_index(drop=True, inplace=True)
```

```python
unique_values_df.to_csv('Values_meanings.csv', index=False)
```
